```python
import tensorflow as tf
from tensorflow.keras.datasets import mnist
from tensorflow.keras.utils import to_categorical
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Flatten
import numpy as np

# Load dataset
(X_train, y_train), (X_test, y_test) = mnist.load_data()

# Normalize
X_train = X_train / 255.0
X_test = X_test / 255.0

# Flatten
X_train = X_train.reshape(-1, 784)
X_test = X_test.reshape(-1, 784)

# One-hot encoding
y_train = to_categorical(y_train, 10)
y_test = to_categorical(y_test, 10)

# Model
model = Sequential([
    Dense(10, activation='softmax', input_shape=(784,))
])

model.compile(optimizer='adam',
              loss='categorical_crossentropy',
              metrics=['accuracy'])

model.summary()

model.fit(X_train, y_train, epochs=10, batch_size=64)

test_loss, test_acc = model.evaluate(X_test, y_test)
print("Test accuracy:", test_acc)
```

no. of layers are increased with the neurons as increasing as 10 , 32 ,64 , 128

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Flatten, Input
from tensorflow.keras.utils import to_categorical
from tensorflow.keras.datasets import mnist

# Load and preprocess MNIST
(X_train, y_train), (X_test, y_test) = mnist.load_data()
X_train, X_test = X_train / 255.0, X_test / 255.0
y_train = to_categorical(y_train, 10)
y_test = to_categorical(y_test, 10)

# parameters requested:
# Neurons: 32, 64, 128 | Layers: +2 | Batches: 32, 64, 128 | Epochs: 10, 15, 20

model_deep = Sequential([
    Flatten(input_shape=(28, 28)),
    Dense(128, activation='relu'),   # Increased neurons
    Dense(64, activation='relu'),    # Extra layer 1
    Dense(32, activation='relu'),    # Extra layer 2
    Dense(10, activation='softmax')
])

model_deep.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])

# Training with increased epochs and batch size
model_deep.fit(X_train, y_train, epochs=20, batch_size=128, validation_split=0.1)

# Evaluate
test_loss, test_acc = model_deep.evaluate(X_test, y_test)
print(f'Test Accuracy with deeper model: {test_acc:.4f}')
```
