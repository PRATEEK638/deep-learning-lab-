back propagation and gradient descent (and gate )

```python
import numpy as np

# Sigmoid
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

# AND gate data
X = np.array([[0,0],[0,1],[1,0],[1,1]])
y = np.array([0,0,0,1])

learning_rates = [0.001, 0.01, 0.05, 0.1, 0.2, 0.5]
epochs = 10

losses = []
w1_list, w2_list, b_list = [], [], []

print("Learning Rate |   w1     w2     b     Loss")
print("--------------------------------------------")

for lr in learning_rates:
    # reset parameters
    w1, w2 = 0.1, 0.1
    b = -0.1

    for _ in range(epochs):
        total_loss = 0
        for i in range(len(X)):
            x1, x2 = X[i]

            # forward pass
            z = w1*x1 + w2*x2 + b
            y_hat = sigmoid(z)

            # loss (MSE)
            loss = 0.5 * (y[i] - y_hat)**2
            total_loss += loss

            # backpropagation (chain rule)
            dE_dyhat = y_hat - y[i]
            dyhat_dz = y_hat * (1 - y_hat)
            delta = dE_dyhat * dyhat_dz

            # gradient descent
            w1 -= lr * delta * x1
            w2 -= lr * delta * x2
            b  -= lr * delta

    losses.append(total_loss)
    w1_list.append(w1)
    w2_list.append(w2)
    b_list.append(b)

    print(f"{lr:12} | {w1:6.3f} {w2:6.3f} {b:6.3f} {total_loss:7.4f}")
```

graph 

```python
import matplotlib.pyplot as plt

plt.plot(learning_rates, losses, marker='o')
plt.xlabel("Learning Rate")
plt.ylabel("Loss")
plt.title("Learning Rate vs Loss (Gradient Descent)")
plt.grid(True)
plt.show()
```
