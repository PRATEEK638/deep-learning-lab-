Implementation of Perceptron Learning Algorithm for Logic Gates

Aim

To implement a Perceptron Learning Algorithm and study its behavior for AND, OR, and XOR logic gates.

Base Code (and gate)

```python
import numpy as np

# AND gate inputs and outputs
x_list = np.array([[0,0],[0,1],[1,0],[1,1]])
y_list = np.array([0,0,0,1])

epochs = 4
w, b = np.zeros(2), 0

for _ in range(epochs):
    for x, y in zip(x_list, y_list):
        z = np.dot(x, w) + b
        y_pred = 1 if z >= 0 else 0
        E = y - y_pred
        print("error:", E, "epoch:", _)
        w += E * x
        b += E
        print("weights:", w, "bias:", b)
```

Increasing epochs improves convergence speed but does not change the final learned weights after convergence.

Task 2: Change Random Weights and Bias
Observation

Random initialization still leads to convergence for linearly separable problems like AND gate.

```python
import numpy as np

# AND gate (Input / Output)
x_list = np.array([[0,0],[0,1],[1,0],[1,1]])
y_list = np.array([0,0,0,1])

# Task 2: Random weights and bias
epochs = 10
w = np.random.rand(2)   # random weights
b = np.random.rand()    # random bias

print("Initial Weights:", w)
print("Initial Bias:", b)

for epoch in range(epochs):
    for x, y in zip(x_list, y_list):
        z = np.dot(x, w) + b
        y_pred = 1 if z >= 0 else 0
        E = y - y_pred
        w += E * x
        b += E
        print("Epoch:", epoch, "Error:", E, "Weights:", w, "Bias:", b)
```

Task 3: Stop Training When Error Becomes Zero
Logic

Training stops if no error occurs for all inputs in an epoch.

```python
import numpy as np

x_list = np.array([[0,0],[0,1],[1,0],[1,1]])
y_list = np.array([0,0,0,1])

w, b = np.zeros(2), 0
epochs = 20

for epoch in range(epochs):
    total_error = 0
    for x, y in zip(x_list, y_list):
        z = np.dot(x, w) + b
        y_pred = 1 if z >= 0 else 0
        E = y - y_pred
        total_error += abs(E)
        w += E * x
        b += E

    print("Epoch:", epoch, "Total Error:", total_error)

    if total_error == 0:
        print("Training stopped – No error")
        break
```
